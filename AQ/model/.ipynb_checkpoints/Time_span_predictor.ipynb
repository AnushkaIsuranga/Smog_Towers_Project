# ARIMA-based Multi-day AQI Forecasting
This notebook demonstrates how to use ARIMA to forecast AQI for the next 3 days using historical AQI data. The approach is univariate (using only the AQI column), but can be extended to multivariate with more advanced models.

```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
import numpy as np

# Load the dataset (update path as needed)
df = pd.read_csv(r'C:/Users/USER/Desktop/AQ/model/dataset/AQI_Dataset.csv', encoding='latin-1')

# Use only the AQI column for ARIMA
aqi_series = df['AQI']

# Split into train and test (last 3 days for test)
train = aqi_series[:-3]
test = aqi_series[-3:]

# Fit ARIMA model (order can be tuned)
model = ARIMA(train, order=(2,1,2))
model_fit = model.fit()

# Forecast next 3 days
yhat = model_fit.forecast(steps=3)

# Print predictions and actuals
print('ARIMA 3-day forecast:', yhat.values)
print('Actual last 3 days:', test.values)
print('RMSE:', np.sqrt(mean_squared_error(test, yhat)))

# Plot
plt.figure(figsize=(10,5))
plt.plot(aqi_series.index, aqi_series, label='Actual AQI')
plt.plot(range(len(aqi_series)-3, len(aqi_series)), yhat, 'ro-', label='ARIMA Forecast (3 days)')
plt.legend()
plt.title('ARIMA 3-Day AQI Forecast')
plt.show()
```
